### 6.1 Listas: `list`
Java necesita `ArrayList<Double> saldos = new ArrayList<>();`. En Python basta con corchetes, y una lista puede mezclar tipos.

````python
saldos = [100.0, 250.5, 80.0]

saldos.append(500.0)        # agregar al final  (Java: add)
saldos.insert(0, 20.0)      # insertar en una posición
print(saldos)

print(saldos[1], saldos[-1])  # acceso por índice
saldos[0] = 25.0              # modificar (Java: set)
saldos.remove(80.0)           # eliminar por valor
print(saldos)

print(len(saldos), sum(saldos), max(saldos), min(saldos))
print(sorted(saldos))         # nueva lista ordenada
print(250.5 in saldos)        # Java: contains
````

````python
Estudiantes = ["Karen", "Domenica", "Kevin", "Pedro", "Andres"]

Estudiantes.append("Jose")
print (Estudiantes)

Estudiantes.insert (1, "Alex")
print (Estudiantes)

Estudiantes.remove("Jose")
print (Estudiantes)

print (Estudiantes [4])

print ("Numero de elementos de la lista: ", len(Estudiantes))

print (Estudiantes[::-1])

print (sorted(Estudiantes))

````

El *slicing* también funciona con listas:

````python
meses = ["ene", "feb", "mar", "abr", "may", "jun"]
print(meses[:3])     # primer trimestre
print(meses[3:])     # segundo trimestre
print(meses[::-1])   # invertida
````

### 6.3 Diccionarios: `dict`
Un diccionario guarda pares **clave → valor**, como un `HashMap<String, Object>` de Java.

````python
socio = {
    "cedula": "0102030405",
    "nombre": "Ana Pérez",
    "saldo": 1520.75,
    "activo": True,
}

print(socio["nombre"])              # Java: map.get("nombre")
socio["saldo"] += 100               # modificar
socio["telefono"] = "0991234567"    # agregar una clave nueva (Java: put)
print(socio.get("correo", "sin correo"))  # valor por defecto si no existe

print(socio.keys())
print("cedula" in socio)            # Java: containsKey
````

````python
estudiante = {
    "nombre": "Santiago Brito",
    "cedula": "01074343227",
    "edad": 20,
    "direccion":"Cuenca",
    "notas": [7.5, 8.0, 9.0, 8.5, 7.3],
    "correo": "arielbrito824@gmail.com"
    } 

print (estudiante["nombre"])
print (estudiante.keys())

print (estudiante["notas"][3])
print (estudiante["direccion"])
````

### 6.4 Conjuntos: `set`
Un conjunto **no admite repetidos**, como el `HashSet` de Java. Es útil para eliminar duplicados.

````python
tipos_cuenta = ["ahorros", "corriente", "ahorros", "plazo fijo", "ahorros"]
unicos = set(tipos_cuenta)
print(unicos, len(unicos))

cuentas_ana = {"ahorros", "plazo fijo"}
cuentas_luis = {"ahorros", "corriente"}
print(cuentas_ana & cuentas_luis)   # intersección: lo que tienen en común
print(cuentas_ana | cuentas_luis)   # unión
````

## 7. Mutabilidad y referencias (10 min)

| Inmutables (no cambian) | Mutables (sí cambian) |
|---|---|
| `int`, `float`, `bool`, `str`, `tuple` | `list`, `dict`, `set` |

Igual que con los objetos en Java, **una variable guarda una referencia**. Si dos variables apuntan a la misma lista, un cambio hecho con una se ve con la otra:

````python
cuentas_a = ["AH-001", "AH-002"]
cuentas_b = cuentas_a          # NO es una copia: es la misma lista
cuentas_b.append("AH-003")
print(cuentas_a)               # ¡también cambió!

copia = cuentas_a.copy()       # copia independiente
copia.append("AH-004")
print(cuentas_a, copia)
````

`==` compara el **contenido** (como `equals` en Java) e `is` compara si es **el mismo objeto** (como `==` con objetos en Java):

````python
x = [1, 2, 3]
y = [1, 2, 3]
print(x == y)   # True: mismo contenido
print(x is y)   # False: son dos listas distintas
````

Vamos a representar a un socio de la cooperativa usando **todos los tipos vistos**:

````python
socio = {
    "cedula": "0102030405",                 # str: puede empezar con 0
    "nombre": "  ana lucía pérez ",          # str con espacios y minúsculas
    "edad": 34,                              # int
    "activo": True,                          # bool
    "garante": None,                         # None: todavía no tiene
    "cuentas": [                             # list de dict
        {"numero": "AH-001", "tipo": "ahorros", "saldo": 1520.75},
        {"numero": "PF-010", "tipo": "plazo fijo", "saldo": 5000.0},
        {"numero": "AH-002", "tipo": "ahorros", "saldo": 80.5},
    ],
}

nombre = socio["nombre"].strip().title()
saldos = [c["saldo"] for c in socio["cuentas"]]
tipos = {c["tipo"] for c in socio["cuentas"]}

print(f"Socio: {nombre} ({socio['cedula']})")
print(f"Cuentas: {len(socio['cuentas'])} | Tipos: {sorted(tipos)}")
print(f"Saldo total: ${sum(saldos):,.2f}")
print(f"Garante: {socio['garante'] or 'sin registrar'}")


#ACTIVIDAD 1

# 1. Obtener el nombre del socio eliminando los espacios innecesarios
nombre_limpio = socio["nombre"].strip()
print(f"Nombre sin espacios: {nombre_limpio}")

# 2. Convertir el nombre a formato de título
nombre_titulo = nombre_limpio.title()
print(f"Nombre en formato título: {nombre_titulo}")

# 3. Mostrar la cédula y la edad
print(f"Cédula: {socio['cedula']} | Edad: {socio['edad']} años")

# 4. Comprobar si el socio está activo
print(f"¿Socio activo?: {socio['activo']}")


#ACTIVIDAD 2 
# 1. Obtener una lista con los saldos de las cuentas
saldos = [c["saldo"] for c in socio["cuentas"]]
print(f"Lista de saldos: {saldos}")

# 2. Obtener una lista con los números de cuenta
numeros = [c["numero"] for c in socio["cuentas"]]
print(f"Números de cuenta: {numeros}")

# 3. Identificar los tipos de cuentas sin valores duplicados utilizando set
tipos = set([c["tipo"] for c in socio["cuentas"]])
print(f"Tipos de cuenta sin duplicados: {tipos}")

# 4. Determinar cuántas cuentas tiene el socio utilizando len()
print(f"Cantidad de cuentas: {len(socio['cuentas'])}")


#ACTIVIDAD 3

# 1. Calcular el saldo total utilizando sum()
saldo_total = sum(saldos)
print(f"Saldo total: ${saldo_total:,.2f}")

# 2. Identificar el saldo máximo utilizando max()
print(f"Saldo máximo: ${max(saldos):,.2f}")

# 3. Identificar el saldo mínimo utilizando min()
print(f"Saldo mínimo: ${min(saldos):,.2f}")

# 4. Calcular el saldo promedio utilizando sum() y len()
promedio = sum(saldos) / len(saldos)
print(f"Saldo promedio: ${promedio:,.2f}")

#ACTIVIDAD 4
# 1. Comprobar si existe un garante registrado
print(f"¿Existe garante registrado?: {socio['garante'] is not None}")
print(f"Garante: {socio['garante'] or 'sin registrar'}")

# 2. Mostrar un mensaje dependiendo del estado del socio
estado_mensaje = "Socio Activo" if socio["activo"] else "Socio Inactivo"
print(f"Estado de {nombre_titulo}: {estado_mensaje}")

# 3. Verificar si el saldo total es mayor a $1.000
print(f"¿El saldo total es mayor a $1.000?: {saldo_total > 1000}")

# 4. Determinar si existe al menos una cuenta de plazo fijo
print(f"¿Tiene cuenta de plazo fijo?: {'plazo fijo' in tipos}")
````

| Entero / decimal | `int`, `long` / `double` | `int` (sin límite) / `float` |
| Sin valor | `null` | `None` |
| Texto | `String`, `char` | `str` (no hay `char`) |
| Dinero exacto | `BigDecimal` | `Decimal` |
| Subcadena | `substring(0, 4)` | `s[0:4]` |
| Lista dinámica | `ArrayList` | `list` → `[ ]` |
| Mapa | `HashMap` | `dict` → `{clave: valor}` |
| Conjunto | `HashSet` | `set` → `{ }` |
| Tipo de un objeto | `instanceof` | `isinstance()` |

````python
socios = [
    {
        "cedula": "0107434322",
        "nombre": "  santiago brito ",
        "cuentas": [
            {"numero": "AH-001", "tipo": "ahorros", "saldo": 1520.75},
            {"numero": "PF-010", "tipo": "plazo fijo", "saldo": 5000.00}
        ]
    },
    {
        "cedula": "0102030405",
        "nombre": "  ana lucía pérez ",
        "cuentas": [
            {"numero": "AH-002", "tipo": "ahorros", "saldo": 850.50},
            {"numero": "CO-001", "tipo": "corriente", "saldo": 1200.00},
            {"numero": "AH-003", "tipo": "ahorros", "saldo": 320.25}
        ]
    },
    {
        "cedula": "0105060708",
        "nombre": "  karen alvarado ",
        "cuentas": [
            {"numero": "AH-004", "tipo": "ahorros", "saldo": 450.00}
        ]
    }
]

nombre_0 = socios[0]["nombre"].strip().title()
cuentas_0 = len(socios[0]["cuentas"])
total_0 = sum([c["saldo"] for c in socios[0]["cuentas"]])

nombre_1 = socios[1]["nombre"].strip().title()
cuentas_1 = len(socios[1]["cuentas"])
total_1 = sum([c["saldo"] for c in socios[1]["cuentas"]])

nombre_2 = socios[2]["nombre"].strip().title()
cuentas_2 = len(socios[2]["cuentas"])
total_2 = sum([c["saldo"] for c in socios[2]["cuentas"]])

print(f"|{'Nombre':<20}|{'Cuentas':>10}|{'Saldo Total':>15}|")
print(f"|{'-'*20}|{'-'*10}|{'-'*15}|")
print(f"|{nombre_0:<20}|{cuentas_0:>10}|${total_0:>14,.2f}|")
print(f"|{nombre_1:<20}|{cuentas_1:>10}|${total_1:>14,.2f}|")
print(f"|{nombre_2:<20}|{cuentas_2:>10}|${total_2:>14,.2f}|")
````

## Soluciones de los ejercicios

````python
# Ejercicio 1
dato = "  0102030405 ; MARÍA josé torres ; 250.50  "
cedula, nombre, saldo = [p.strip() for p in dato.split(";")]
nombre = nombre.title()
saldo = float(saldo) + 49.50
print(nombre)
print(f"Saldo: ${saldo:.2f}")
print("Cédula válida:", cedula.isdigit() and len(cedula) == 10)
````

````python
# Ejercicio 2
depositos = [120.0, 45.5, 300.0, 45.5, 80.0]
depositos.append(200.0)
total = sum(depositos)
print(f"Total: ${total:.2f}")
print(f"Promedio: ${total / len(depositos):.2f}")
print("Mayor:", max(depositos), "| Menor:", min(depositos))
print("Montos distintos:", len(set(depositos)))
````

````python
# Ejercicio 3
tasas = {"ahorros": 0.02, "plazo fijo": 0.07, "corriente": 0.0}
print(f"Plazo fijo: {tasas['plazo fijo']:.1%}")
tasas["infantil"] = 0.03
print(tasas.get("vivienda", 0.0))
print(tasas)
````
